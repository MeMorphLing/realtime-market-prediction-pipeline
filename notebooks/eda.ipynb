# Market Prediction — Exploratory Data Analysis

This notebook walks through the raw ingested data, sentiment distributions, and feature windows used to train the sequential models.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

RAW_PRICES = Path('../data/raw/prices')
RAW_NEWS = Path('../data/raw/news')
RAW_SOCIAL = Path('../data/raw/social')
FEATURES = Path('../data/features')

sns.set_theme(style='darkgrid')
pd.options.display.max_columns = 30
```

## 1. Price data

```python
files = sorted(RAW_PRICES.glob('*.parquet'))
print(f'Found {len(files)} price files')
if files:
    df = pd.read_parquet(files[-1])
    display(df.head())
    df['close'].plot(title='Latest close prices')
    plt.show()
```

## 2. News headlines

```python
files = sorted(RAW_NEWS.glob('*.parquet'))
if files:
    news = pd.read_parquet(files[-1])
    display(news.head())
    print('Total articles:', len(news))
```

## 3. Social media volume

```python
files = sorted(RAW_SOCIAL.glob('*.parquet'))
for f in files[-2:]:
    df = pd.read_parquet(f)
    print(f.name, '->', len(df), 'rows')
    display(df.head())
```

## 4. Feature windows

```python
files = sorted(FEATURES.glob('*_windows.parquet'))
if files:
    f = files[-1]
    windows = pd.read_parquet(f)
    print(f.name, '->', len(windows), 'windows')
    display(windows.head(3))
```
